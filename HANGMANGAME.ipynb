```python
word_categories = {
"1": { "name": "Animal Kingdom", 
        "words": ["rhinoceros", "giraffe", "platypus", "leopard", "swordfish", "octopus"]}, 
"2": { "name": "Food", "words": ["spaghetti", "chocolate", "beetroot", "cauliflower", "broccoli", "paella"]}, 
"3": { "name": "Music", "words": ["melody", "pitch", "harmonica", "saxophone", "accordion", "violin"]}
    }
       
```

```python
print( "Choose a topic:")
for number, category in word_categories.items():
    print(number, "-", category["name"])
```

```python
choice= input("Choose a topic:")
```

```python
if choice in word_categories:
    print("You chose:", word_categories[choice]["name"])
else:
    print("Invalid choice. Please choose the listed categories")
```

```python
selected_words = word_categories[choice]["words"]
```

```python
import random
```

```python
secret_word = random.choice(selected_words)
```

```python
game_state = {"lives": 5, "guessed_letters": [], "game_over": False}
```

```python
def display_word(secret_word, guessed_letters):
    display = []

    for letter in secret_word:
        if letter in guessed_letters:
            display.append(letter)
        else:
            display.append("_")

    return " ".join(display)
```

```python
guess = input("Choose your letter: ").lower()

if guess in secret_word:
    print("Correct!")
else:
    print("Wrong!")
    game_state["lives"] -= 1

if game_state["lives"] <= 0:
    game_state["game_over"] = True
    print("Game over!")
```

```python
if len(guess) != 1 or not guess.isalpha():
    print("Please enter one letter.")
elif guess in game_state["guessed_letters"]:
    print("That letter has already been chosen.")
else:
    game_state["guessed_letters"].append(guess)

    if guess in secret_word:
        print("Correct!")
    else:
        print("Wrong!")
        game_state["lives"] -= 1
        
print(display_word(secret_word, game_state["guessed_letters"]))
```

```python
while not game_state["game_over"]:

    guess = input("Choose your letter: ").lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter ONE letter.")

    elif guess in game_state["guessed_letters"]:
        print("You already guessed that letter.")

    else:
     game_state["guessed_letters"].append(guess)

    if guess in secret_word:
        print("Correct!")
    else:
        print("Wrong!")
        game_state["lives"] -= 1

    print("Lives:", game_state["lives"])
    print(display_word(secret_word, game_state["guessed_letters"]))

    if game_state["lives"] <= 0:
        game_state["game_over"] = True
        print("Game over!")

    if "_" not in display_word(secret_word, game_state["guessed_letters"]):
        game_state["game_over"] = True
        print("You won!")
```
